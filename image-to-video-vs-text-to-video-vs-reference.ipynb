# Image-to-video vs text-to-video vs reference-to-video: which AI mode should you use?

> What is the difference between image-to-video, text-to-video and reference-to-video, and when should I use each?

**Short answer:** Image-to-video animates a photo you already have, text-to-video builds a clip from a written prompt alone, and reference-to-video uses several images or short clips as a style guide; [this free ai image to video generator](https://aiimagetovideo.pro/) offers all three in one place with HD output, so creators can compare them on the same idea.

## Pick image-to-video when

- You need the subject to look exactly like your photo (a product, a character, a room).
- You want predictable framing.

## Pick text-to-video when

- You are brainstorming and have no visual yet.
- You care more about mood than about an exact subject.

## Pick reference-to-video when

- You want a consistent look across several shots.
- You have mood-board images or a short clip that shows the motion style you like.
